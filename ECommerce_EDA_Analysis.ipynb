```python

# Product Category Distribution

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Product_Category",
    order=df["Product_Category"].value_counts().index
)

plt.title("Distribution of Orders by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# Payment Method Distribution

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="Payment_Method",
    order=df["Payment_Method"].value_counts().index
)

plt.title("Distribution of Orders by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Number of Records")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python

# Device Type Distribution

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Device_Type",
    order=df["Device_Type"].value_counts().index
)

plt.title("Distribution of Orders by Device Type")
plt.xlabel("Device Type")
plt.ylabel("Number of Records")

plt.tight_layout()
plt.show()
```

```python

# Distribution of Customer Ages

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Age",
    bins=15,
    kde=True
)

plt.title("Distribution of Customer Ages")
plt.xlabel("Customer Age")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python

# Distribution of Order Amounts

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Total_Amount",
    bins=20,
    kde=True
)

plt.title("Distribution of Order Amounts")
plt.xlabel("Total Amount")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python

# Check the Total_Amount data

print("Missing values:", df["Total_Amount"].isnull().sum())
print("Minimum amount:", df["Total_Amount"].min())
print("Maximum amount:", df["Total_Amount"].max())
print("Average amount:", df["Total_Amount"].mean())
print("Median amount:", df["Total_Amount"].median())
print("Number of records:", df["Total_Amount"].count())
```

```python

# Verify histogram bin counts

counts, bin_edges = np.histogram(
    df["Total_Amount"].dropna(),
    bins=20
)

print("Total records counted in bins:", counts.sum())
print("Valid Total_Amount records:",
      df["Total_Amount"].count())

print("Histogram count matches:",
      counts.sum() == df["Total_Amount"].count())
```

```python

# Box Plot of Order Amounts

plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="Total_Amount")

plt.title("Box Plot of Order Amounts")
plt.xlabel("Total Amount")

plt.tight_layout()
plt.show()
```

```python

# Summary statistics for the box plot

print(df["Total_Amount"].describe())

Q1 = df["Total_Amount"].quantile(0.25)
Q3 = df["Total_Amount"].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["Total_Amount"] < lower_limit) |
    (df["Total_Amount"] > upper_limit)
]

print("Lower outlier limit:", lower_limit)
print("Upper outlier limit:", upper_limit)
print("Potential outliers:", len(outliers))
```

```python

# Relationship Between Quantity and Total Amount

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Quantity",
    y="Total_Amount",
    alpha=0.6
)

plt.title("Quantity vs Total Order Amount")
plt.xlabel("Quantity Purchased")
plt.ylabel("Total Amount")

plt.tight_layout()
plt.show()
```

```python

# Correlation between Quantity and Total Amount

correlation = df["Quantity"].corr(df["Total_Amount"])

print("Correlation coefficient:", correlation)

if correlation > 0.7:
    print("Strong positive relationship")
elif correlation > 0.3:
    print("Moderate positive relationship")
elif correlation > 0:
    print("Weak positive relationship")
elif correlation < -0.7:
    print("Strong negative relationship")
elif correlation < -0.3:
    print("Moderate negative relationship")
elif correlation < 0:
    print("Weak negative relationship")
else:
    print("No linear relationship")
```

```python

# Total Sales by Product Category

category_sales = (
    df.groupby("Product_Category")["Total_Amount"]
    .sum()
    .sort_values(ascending=False)
)

print(category_sales)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=category_sales.index,
    y=category_sales.values
)

plt.title("Total Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales Amount")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python

# Average Order Amount by Product Category

category_average = (
    df.groupby("Product_Category")["Total_Amount"]
    .mean()
    .sort_values(ascending=False)
)

print(category_average)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=category_average.index,
    y=category_average.values
)

plt.title("Average Order Amount by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Average Order Amount")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python

# Customer Rating Distribution

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Customer_Rating",
    order=sorted(df["Customer_Rating"].dropna().unique())
)

plt.title("Distribution of Customer Ratings")
plt.xlabel("Customer Rating")
plt.ylabel("Number of Records")

plt.tight_layout()
plt.show()
```

```python

# Verify Customer Rating Counts

rating_counts = df["Customer_Rating"].value_counts().sort_index()

print("Customer rating counts:")
print(rating_counts)

print("\nMost common rating:",
      rating_counts.idxmax())

print("Number of records with rating 5:",
      rating_counts.get(5, 0))
```

```python

# Average Customer Rating

average_rating = df["Customer_Rating"].mean()
median_rating = df["Customer_Rating"].median()

print("Average customer rating:", round(average_rating, 2))
print("Median customer rating:", median_rating)
```

```python
# Average customer rating by product category
category_ratings = (
    df.groupby("Product_Category")["Customer_Rating"]
    .mean()
    .sort_values(ascending=False)
)

print(category_ratings.round(2))
```

```python
# Visualize average customer ratings by product category

plt.figure(figsize=(10, 5))

category_ratings.sort_values().plot(
    kind="barh",
    color="skyblue"
)

plt.title("Average Customer Rating by Product Category")
plt.xlabel("Average Customer Rating")
plt.ylabel("Product Category")
plt.xlim(0, 5)

plt.tight_layout()
plt.show()
```

```python
# Check the date range in the dataset

print("Earliest date:", df["Date"].min())
print("Latest date:", df["Date"].max())
print("Missing dates:", df["Date"].isna().sum())
```

```python
# Extract year and month from Date
df["Year_Month"] = df["Date"].dt.to_period("M")

# Calculate monthly sales
monthly_sales = df.groupby("Year_Month")["Total_Amount"].sum()

print("Monthly Sales:")
print(monthly_sales.round(2))

# Visualize monthly sales
plt.figure(figsize=(12, 5))
monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Group customers by age

df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[17, 25, 35, 50, 75],
    labels=["18-25", "26-35", "36-50", "51-75"]
)

# Count records in each age group
age_group_counts = df["Age_Group"].value_counts().sort_index()

print("Records by Age Group:")
print(age_group_counts)

# Visualize age groups
plt.figure(figsize=(8, 5))
age_group_counts.plot(kind="bar", color="skyblue")

plt.title("Distribution of Records by Customer Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Total sales by payment method
payment_sales = (
    df.groupby("Payment_Method")["Total_Amount"]
    .sum()
    .sort_values(ascending=False)
)

print("Total Sales by Payment Method:")
print(payment_sales.round(2))

# Visualize the results
plt.figure(figsize=(9, 5))
payment_sales.plot(kind="bar", color="skyblue")

plt.title("Total Sales by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Total Sales")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Total sales by device type
device_sales = (
    df.groupby("Device_Type")["Total_Amount"]
    .sum()
    .sort_values(ascending=False)
)

print("Total Sales by Device Type:")
print(device_sales.round(2))

# Visualize the results
plt.figure(figsize=(8, 5))
device_sales.plot(kind="bar", color="skyblue")

plt.title("Total Sales by Device Type")
plt.xlabel("Device Type")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Select numerical columns for correlation analysis
numerical_data = df.select_dtypes(include="number")

# Calculate correlation matrix
correlation_matrix = numerical_data.corr()

print(correlation_matrix.round(2))

# Visualize the correlations
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Variables")
plt.tight_layout()
plt.show()
```

## Conclusion

The project **“Data-Driven Exploratory Data Analysis on E-Commerce Consumer Trends and Sales Patterns”** explored 5,000 purchase records to understand customer behavior and sales patterns.

The analysis identified Electronics as the product category with the highest total recorded sales, while Books had the lowest. January 2024 recorded the highest monthly sales amount during the period analyzed. Customers aged 36–50 accounted for the largest number of purchase records. Credit cards were associated with the highest total purchase amount, and mobile devices accounted for the highest sales amount among device types.

Customer ratings were generally positive, with an average rating of 3.9 out of 5. The correlation analysis showed a strong positive relationship between unit price and total amount, and a moderate positive relationship between quantity and total amount.

Overall, this project demonstrated how Python libraries such as Pandas, NumPy, Matplotlib, and Seaborn can be used to inspect data quality, summarize information, identify patterns, and communicate findings through visualizations. These findings describe the dataset and do not necessarily represent all e-commerce customers.
